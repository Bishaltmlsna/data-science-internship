#Create a list of numbers from 1 to 20 and extract all prime numbers.

```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, n):
        if n % i == 0:
            return False
    return True
primes = []
for num in range(2, 21):
    if is_prime(num):
        primes.append(num)
print(primes)
```
